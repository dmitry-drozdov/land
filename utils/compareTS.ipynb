```python
import os
from collections import Counter
```

```python
def read_file_lines(filepath):
    """Считывает строки из файла и возвращает их в виде словаря с количеством вхождений."""
    if not os.path.exists(filepath):
        return None
    with open(filepath, 'r', encoding='utf-8') as file:
        return Counter(line.strip() for line in file)
```

```python
def compare_files(directory):
    total = 0
    ok = 0
    """Рекурсивно обходит файлы в указанной папке и сравнивает пары файлов .res и .res_land."""
    for root, _, files in os.walk(directory):
        for filename in files:
            if filename.endswith(".res") and not filename.endswith(".res_land"):
                base_name = filename[:-4]  # Убираем расширение .res
                res_file = os.path.join(root, filename)
                res_land_file = os.path.join(root, f"{base_name}.res_land")
                
                lines_res = read_file_lines(res_file)
                lines_res_land = read_file_lines(res_land_file)
                
                total+=1
                
                if lines_res is None or lines_res_land is None:
                    print(f"Файл отсутствует: {res_file if lines_res is None else res_land_file}")
                    continue
                
                if lines_res == lines_res_land:
                    ok += 1
                    #print(f"✔ Файлы {filename} и {base_name}.res_land совпадают")
                else:
                    missing_in_res = {line: count for line, count in lines_res_land.items() if count > lines_res.get(line, 0)}
                    missing_in_res_land = {line: count for line, count in lines_res.items() if count > lines_res_land.get(line, 0)}
                    print(f"❌ Разница в файлах {root} {filename} и {base_name}.res_land:")
                    if missing_in_res:
                        print("  В .res_land есть, а в .res не хватает:", missing_in_res)
                    if missing_in_res_land:
                        print("  В .res есть, а в .res_land не хватает:", missing_in_res_land)
    print(f"OK: {ok} ({ok/total*100}%), BAD: {total-ok}, TOTAL: {total}")
```

```python
directory_path = "e:\\phd\\ts_res"
compare_files(directory_path)
```
